```python
import os
import glob
import numpy as np
import pandas as pd

root = "/home/boat/proxyISP/pytorch-superpoint/logs/"
folder_pattern = (
    "eval_ll_CMAES_train_lowlight_maxstd0.01_csadampfac10.0_*_HPatchesV4.1"
)

results = []

pattern = os.path.join(root, folder_pattern, "predictions", "result.npz")
print("pattern", pattern)

for path in glob.glob(pattern):
    folder = os.path.basename(os.path.dirname(os.path.dirname(path)))[-22:]

    data = np.load(path)
    correctness = data["correctness"]  # shape (N, 6)
    repeatability = data.get("repeatability")
    mscore = data.get("mscore")
    mAP = data.get("mAP")
    loc_err = data.get("localization_err")  # NEW

    # Threshold averages
    thresh_means = correctness.mean(axis=0)
    avg_overall = thresh_means.mean()

    # Compute mean metrics
    mean_repeat = np.mean(repeatability) if repeatability is not None else np.nan
    mean_mscore = np.mean(mscore) if mscore is not None else np.nan
    mean_map = np.mean(mAP) if mAP is not None else np.nan
    mean_loc_err = np.mean(loc_err) if loc_err is not None else np.nan  # NEW

    results.append(
        (folder, *thresh_means.tolist(), avg_overall, mean_repeat, mean_mscore, mean_map, mean_loc_err)
    )

columns = ["folder", "t1", "t2", "t3", "t4", "t5", "t6", "avg",
           "mean_repeat", "mean_mscore", "mean_map", "mean_loc_err"]
df = pd.DataFrame(results, columns=columns)

# sort by avg correctness
df = df.sort_values(by="mean_loc_err", ascending=False).reset_index(drop=True)

df
```
